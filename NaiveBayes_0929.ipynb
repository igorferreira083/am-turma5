# Algoritmo e Naive Bayes [Credit_Data]


```python
from sklearn.naive_bayes import GaussianNB
```

# Carregamento de Dados Pickle

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pickle
with open('/content/drive/MyDrive/am-t6/credit.pkl', 'rb') as f:
    X_credit_train, y_credit_train, X_credit_test, y_credit_test = pickle.load(f)
```

```python
X_credit_train.shape, y_credit_train.shape
```

```python
X_credit_test.shape, y_credit_test.shape
```

# Montagem de Neurônio Especial

```python
neuronio_nb = GaussianNB()
neuronio_nb.fit (X_credit_train, y_credit_train)
```

Cenário Sintético

```python
sintetico_credito = neuronio_nb.predict(X_credit_test)
```

```python
sintetico_credito
```

```python
y_credit_test
```

# Testes de Acurácia do Modelo

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_credit_test, sintetico_credito)
```

Escala Matriarcal de Acertos e Erros

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(neuronio_nb)
cm.fit(X_credit_train, y_credit_train)
cm.score(X_credit_test, y_credit_test)
```

```python
print(classification_report(y_credit_test, sintetico_credito))
```
